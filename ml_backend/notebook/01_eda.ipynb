```python
import pandas as pd 
import numpy as np
from src.utility import get_logger, config, CustomException
```

```python
data_path = config.load('data_source')['data']['data_source']
project_root = config.config_dir.parent
data_path = project_root / data_path
```

```python
df = pd.read_csv(data_path)
```

```python
df.head()
```

```python
df.columns = df.columns.str.lower()
```

```python
df.head()
```

```python
status_values = {1:'ok',
                2: 'default',
                0 : 'unk'
               } 

df.status = df.status.map(status_values)
```

```python
df.head()
```

```python
home_vale = {
    1:'rent',
    2: 'owner',
    3: 'private',
    4: 'ignore',
    5 : 'parents',
    6 : 'other',
    0 : 'unk'
}
df.home = df.home.map(home_vale)
```

```python
marital_values = {
    1:'single',
    2: 'married',
    3: 'widow',
    4: 'separated',
    5 : 'divorced',
    6 : 'other',
    0 : 'unk'
}
df.marital = df.marital.map(marital_values)
```

```python
records_values = {
    1:'no',
    2: 'yes',
    0 : 'unk'
}
df.records = df.records.map(records_values)
```

```python
job_values = {
    1:'fixed',
    2: 'partime',
    3: 'freelance',
    4:'others',
    0 : 'unk'
}
df.job = df.job.map(job_values)
```

```python
df.head()
```

```python
df.describe()
```

```python
df.income.max()
```

```python
for c in ['income', 'assets', 'debt']:
    df[c] = df[c].replace(to_replace=99999999, value=np.nan)
print("New max income:", df['income'].max())
```

```python
df.income.max()
```

```python
df.status.value_counts()
```

```python
df = df[df.status != 'unk'].reset_index(drop=True)
```

```python
df.status.value_counts()
```

```python
df.head()
```

```python
from sklearn.model_selection import train_test_split 
```

```python
df_full_train, df_test = train_test_split(df,test_size = 0.2,random_state=40)
df_train, df_val = train_test_split(df_full_train,test_size=0.25,random_state=40)
```

```python
df_train = df_train.reset_index(drop=True)
df_val = df_val.reset_index(drop=True)
df_test = df_test.reset_index(drop=True)
```

```python
df_train
```

```python
y_train= (df_train.status =='default').astype('int').values
y_val= (df_val.status =='default').astype('int').values
y_test= (df_test.status =='default').astype('int').values
```

```python
y_train
```

```python
del df_train['status']
del df_val['status']
del df_test['status']
```

```python
df_train.head()
```

```python
from sklearn.tree import DecisionTreeClassifier
from sklearn.feature_extraction import DictVectorizer
from sklearn.metrics import roc_auc_score
```

```python
train_dicts = df_train.fillna(0).to_dict(orient='records')
dv = DictVectorizer(sparse=False)
X_train = dv.fit_transform(train_dicts)
X_train
```

```python
dt = DecisionTreeClassifier()
dt.fit(X_train,y_train)
```

```python
val_dict = df_val.fillna(0).to_dict(orient='records' )
X_val = dv.transform(val_dict)
```

```python
y_pred = dt.predict_proba(X_val)[:,1]
y_pred
```

```python
roc_auc_score(y_val,y_pred)
```

```python
y_predic = dt.predict_proba(X_train)[:,1]
```

```python
roc_auc_score(y_train,y_predic)
```

```python
dt = DecisionTreeClassifier(max_depth = 4)
dt.fit(X_train,y_train)
```

```python
val_dict = df_val.fillna(0).to_dict(orient='records' )
X_val = dv.transform(val_dict)
y_pred = dt.predict_proba(X_val)[:,1]
roc_auc_score(y_val,y_pred)
```
